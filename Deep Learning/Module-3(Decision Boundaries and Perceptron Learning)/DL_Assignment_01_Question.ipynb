# Question 04:

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:
Perceptron's weight update formula is, W*new* = W*old* + n*(y - y_hat)*Xi.

Here when the prediction is wrong. Let, y=0

if, y_hat = 1 the prediction is wrong and so Error = (y-y_hat) = (0-1) = -1. Now according to the formula if the value is set then the weight will be updated, W*new* = W*old* - n*Xi. and the new weight will be for the ground truth.

if, y_hat = 0 the prediction is right and so Error = (y-y_hat) = (0-0) = 0. Now according to the formula if the value is set then the weight will not be updated W*new* = W*old* + 0.
