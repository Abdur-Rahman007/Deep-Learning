## Write Answer 02:


```python
class Model(nn.Module):
  def __init__(self, num_features):
    super().__init__()

    # Hidden layer
    self.linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()

    # Output layer
    self.linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear1(features)
    out = self.relu(out)
    out = self.linear2(out)
    out = self.sigmoid(out)

    return out
```

**Number of hidden neurons** : there are 3 hidden neurons, which are in the hidden layer.

**Choice of activation function** : There are two activation function Sigmoid and ReLU. Here Sigmoid is used in hidden layer and relu is used in output layer.

# Question 04:

## Implement a full training loop:

Forward pass

Loss computation

Backward pass

Parameter update

Gradient reset

### Requirements:

Train for at least 100 epochs.

Print loss every 10 epochs.

Store training loss history(You can pick your own Data Structure).

## Write Answer 04:

```python
epochs = 100
model = Model(X_train_tensor.shape[1])

optimizer = torch.optim.SGD(model.parameters(), lr = learning_rate)

for epoch in range(epochs):
  # forward pass
  y_pred = model(X_train_tensor)

  # Loss computation
  loss = loss_function(y_pred, y_train_tensor.reshape(-1, 1))

  # Zero gradient
  optimizer.zero_grad()

  # backward pass
  loss.backward()

  # parameter update
  optimizer.step()

  #print
  if(epoch%10==0):
    print(f"{epoch}: Loss = {loss.item()}")
```

## Write Answer 07:

Why must gradients be reset every epoch?

**Ans:** Gradients must be reset every epoch because pytorch accumulates it by default. If it is not cleared the gradients will be added with the previous one, which will give wrong calculation. When it is cleared the gradients only operate with the current value.

What happens if learning rate is too high?

**Ans:** The learning rate is declared to minimize the loss in a specific rate. If the learning rate is too high the optimizer may overshoot the minimum loss for taking big steps toward the local minimum.

What happens if learning rate is too small?

**Ans:** If the learning rate is too samll it may take long time to reaching the optimal solution. Or it may stuck before reahing the solution

Why do we define layers inside the constructor (__init__) and not inside forward()?

**Ans:** Layers are defined int the constructors so that the prametrs can be optimized during training. If the layers were define inside the forward method the exact layer can not be accessed because the forward method will create new layers in every forward pass
