## Explain different activation functions used in neural networks, including Sigmoid, Tanh, and ReLU. Provide their mathematical formulas, advantages, disadvantages, and suitable use cases. Also discuss why ReLU is widely used in deep learning.

## Write Answer 06:
Activation functions is used for non-linearity, it allows neural networks to model complex relationships. Without activation function, every deep networks collapse into a single linear transformation.

### 1. Sigmoid Activation
formula : σ(x) = 1/ (1 + e^-x)

The sigmoid function generates a S shaped curve. It's output range is between (0 to 1). It is good for probability outputs and it is also smooth and differentiable. The problems are vanishing gradient, slow convergence(output not zero-centered). It is used mostly for binary classification and logistic regression models.

### 2. Tanh Activation
formula : tanh(x) = (e^x - e^-x) / (e^x + e^-x)

The tanh is zero centered and it's output range is between (-1 to 1). Some advantanges are it is better than sigmoid for hidden layers and it's convergence  is faster (due to zero-centered output). The problem is for this also vanishing gradient. It is used in hidden layers.

### 3. ReLU Activation
formula: f(x) = max(0, x)
The ReLU is a piecewise function and it's output range is between [0, ∞). It's advantages are, it avoids vanishing gradient, computationally efficientand speed up training. Some disadvantages are Dying ReLU problem (neurons stuck at 0), it is not zero-centered and also it can be unstable for very large values. It is used for hidden layers in deep networks and also used in CNNs, deep ANNs, modern architectures.

### ReLU is widely used because-
* Solves Vanishing Gradient
* Computational Efficiency, faster training
* Better Performance in Deep Networks

# Question 08:

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity.

## Write Answer 08:
In Convolutional Neural Networks (CNNs) pooling reduces the size of feature maps while retaining the most important information. It operates on small regions (like 2×2 windows) and summarizes each region into a single value. Among poolings MaxPooling selects max value, it focus on most domain feature, can ignore smallnoise and AveragePooling computes mean values, focus on overall presence of features, it smooths noise but may blur features.
### Advantages and impact of MaxPooling
* Captures important features
* Provides translation invariance
* Works well for object detection and classification

### Advantages and impact of AveragePooling
* Preserves background information
* Produces smoother feature maps
* Less sensitive to outliers than max pooling

### Pooling helps reduce overfitting in several ways:
* Reduces parameters: Smaller feature maps, so fewer weights in fully connected layers.
* Provides invariance: Model becomes less sensitive to small shifts.
* Forces the network to learn robust features, not exact pixel positions.

### Pooling helps Reducing Computational Complexity:
* Shrinks feature map size, so fewer computations in later layers.
* like Input: 32×32 -> After pooling: 16×16

```python
# train test split
X = df.iloc[:, 1:].values
y = df.iloc[:, 0].values
```

## Custom Dataset

```python
class MultiClassClassfication(Dataset):
  def __init__(self, df):
    self.labels = y
    self.images = X

  def __len__(self):
    return len(self.labels)

  def __getitem__(self, idx):
    image = torch.tensor(self.images[idx], dtype=torch.float32)
    image = image.view(1, 28, 28)

    label = torch.tensor(self.labels[idx], dtype=torch.long)

    return image, label
```

## Load Data

```python
dataset = MultiClassClassfication(df)

train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size

train_dataset, test_dataset = random_split(dataset, [train_size, test_size])

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)

''' can also done for the csv dataset
# Split DataFrame
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Create datasets
train_dataset = FashionMNISTDataset(train_df)
test_dataset = FashionMNISTDataset(test_df)

# DataLoaders
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)
'''
```

## CNN Model

```python
class MyCNN(nn.Module):
  def __init__(self, num_classes=10):
    super().__init__()

    self.features = nn.Sequential(
        #block 1
        nn.Conv2d(1 ,32, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(32),
        nn.Conv2d(32, 32, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(32),
        nn.MaxPool2d(2),

        #block 2
        nn.Conv2d(32 , 64, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(64),
        nn.Conv2d(64, 64, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(64),
        nn.MaxPool2d(2),

        #block 3
        nn.Conv2d(64 ,128 , kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(128),
        nn.Conv2d(128, 128, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(128),
        nn.MaxPool2d(2),

        #block 4
        nn.Conv2d(128 ,256 , kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(256),
        nn.Conv2d(256, 256, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.BatchNorm2d(256),
        nn.MaxPool2d(2),
    )

    self.classifier = nn.Sequential(
        nn.Flatten(),
        nn.Linear(256 * 1 * 1, 128),
        nn.ReLU(),
        nn.Dropout(0.4),

        nn.Linear(128, 64),
        nn.ReLU(),
        nn.Dropout(0.4),

        nn.Linear(64, num_classes)
    )


  def forward(self, x):
    x = self.features(x)
    x = self.classifier(x)
    return x
```

```python
model = MyCNN(num_classes=10)
```

```python
criterion = nn.CrossEntropyLoss()  # Loss function
optimizer = optim.Adam(model.parameters(), lr= 0.001)  # Optimizer
```

# Training

```python
epochs = 10

for epoch in range(epochs):
    model.train()
    total_loss = 0

    for images, labels in train_loader:
        outputs = model(images)
        loss = criterion(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1}/{epochs}, Loss: {total_loss/len(train_loader):.4f}")
```

## Evaluation Function

```python
def evaluate(loader):
  model.eval()
  correct = 0
  total = 0

  with torch.no_grad():
      for images, labels in loader:
          outputs = model(images)
          _, predicted = torch.max(outputs, 1)

          total += labels.size(0)
          correct += (predicted == labels).sum().item()

  return correct / total
```

```python
test_acc = evaluate(test_loader)
print("Test Accuracy:", test_acc)
```
