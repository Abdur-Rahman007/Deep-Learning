# Question 01:
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

**Write Answer 01:**

Gradient descent is a iterative optimization algorithm used to minimize the loss or cost functon in machine learning and deep learning. The main idea behnd the geadient descent is to find the local minimum by moving in the opposite direction of the gradient. For minimizing the loss function it calculates the derivative or slope of the loss function with respect of each parameter and predict the error if the error or loss is greater then it updates the weight and bias. The learning rate can be called size of step toward the minimum loss. If the learning rate is too high it will go faster towards the goal in big steps so, it can overshoot the minimum. If the learning rate is too low it will take much time towards the goal using small steps.

# Question 05:
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:
Softmax activation function converts lists of number into a probability number. Where each value is between 0 and 1, the total sum of the values is 1. Softmax is used in multi-class classification problem means where is more than two classes only one class can be correct among them. And the sum of all probability equals 1 makes is perfect for multi-class probability classification. The sigmoid function is used for binary-class classification but softmax is used in multi-class calssification problem.
